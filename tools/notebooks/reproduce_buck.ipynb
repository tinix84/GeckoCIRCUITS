# Reproduce the buck simulation
Run `buck_simple.ipes`, export the simulation as a research bundle, and submit the bundle for verification.

Start the GeckoCIRCUITS REST API first. Generate and install the Python client. See the Research Bundles documentation for setup.


```python
import sys
from pathlib import Path

root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / 'pom-reactor.xml').is_file())
sys.path.insert(0, str(root / 'tools' / 'notebooks'))
from acceptance import configuration, run_buck
from gecko_client import ApiClient

config = configuration()
print(config.host)
```

## Run, export, and verify
A repeat on the same build should give zero deviation for each signal.


```python
with ApiClient(config) as client:
    archive, report = run_buck(client)

assert report.passed
assert all(value == 0 for value in report.max_relative_deviation.values())
print(f'Bundle file: {archive}')
print(f'Verified signals: {len(report.max_relative_deviation)}')
print(f'Maximum deviation: {max(report.max_relative_deviation.values()):.2%}')
```
